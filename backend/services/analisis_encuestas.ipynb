```python
#importa librerias

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import streamlit as st
```

```python
# cargar url con el formulario de google forms en formato csv

url = "https://docs.google.com/spreadsheets/d/e/2PACX-1vQuM2Y9QqcH_1LIkmdHHflDYTK7y3u9jgUlmZ-XYzaky7HYNyN3_DceBNSagYXESgdYZgnKSRIv5CkU/pub?gid=1237250350&single=true&output=csv"

df = pd.read_csv(url)

#limpia las respuestas de espacios en blanco
df["estas feliz"] = df["estas feliz"].str.strip()

#pagina de streamlit
st.title("Resultados de la Encuesta")

```

```python
# Mostrar cuántas respuestas hubo de cada tipo
print(df["estas feliz"].value_counts())
# Mostrar cuántas respuestas hubo de cada tipo
print(df["¿Que porcentaje de felicidad sentis ahora mismo?"].value_counts()) 
```

```python
# Contar respuestas
conteo = df["estas feliz"].value_counts()


# Mostrar tabla
st.write(conteo)


#print(df.head()) #muestra las respuestas del formulario

print(df.columns) #muestra las preguntas del formulario

# Crear gráfico de barras
df["estas feliz"].value_counts().plot(kind="bar")

plt.title("¿Estás feliz?")
plt.xlabel("Respuesta")
plt.ylabel("Cantidad de personas")
plt.gca().yaxis.set_major_locator(plt.MaxNLocator(integer=True))
plt.show()

# Crear gráfico de barras
df["¿Que porcentaje de felicidad sentis ahora mismo?"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("¿Que porcentaje de felicidad sentis ahora mismo?")
plt.ylabel("")  # Oculta la etiqueta del eje
plt.show()
```
